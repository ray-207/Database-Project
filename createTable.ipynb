```python
import pandas as pd
```

```python
df = pd.read_csv("./clean2.csv")
print(df)
```

```python
#customers
customerDf = pd.DataFrame(
    columns=[
        "CustomerID",
        "CountryOrigin"
    ]
)
print(customerDf)
```

```python
customerDf = df[["Customer ID", "Country"]].drop_duplicates(subset="Customer ID")
customerDf.to_csv("customers.csv", index=False)
print(customerDf)
```

```python
#items
itemDf = pd.DataFrame(
    columns=[
        "StockCode",
        "ItemDescription",
        "Price"
    ]
)
print(itemDf)
```

```python
itemDf = df[["StockCode", "Description", "Price"]].drop_duplicates(subset = "StockCode")
itemDf.to_csv("items.csv", index=False)

print(itemDf)
```

```python
#invoice
invoiceDf = pd.DataFrame(
    columns=[
        "InvoiceID",
        "InvoiceDate",
        "customerID"
    ]
)

print(invoiceDf)
```

```python
invoiceDf = df[["Invoice", "InvoiceDate", "Customer ID"]].drop_duplicates(subset="Invoice")
invoiceDf.to_csv("invoice.csv", index=False)
print(invoiceDf)
```

```python
#itemOrder
itemOrderDf = pd.DataFrame(
    columns=[
        "Quantity",
        "InvoiceID",
        "StockCode"
    ]
)

print(itemOrderDf)
```

```python
itemOrderDf = df[["Quantity", "Invoice", "StockCode"]]
itemOrderDf.to_csv("itemOrder.csv", index=False)
print(itemOrderDf)
```
