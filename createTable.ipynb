```python
import pandas as pd
import random
```

```python
df = pd.read_csv("./clean2.csv")
print(df)
```

```python
#customers
customerDf = pd.DataFrame(
    columns=[
        "CustomerID",
        "CountryOrigin"
    ]
)
print(customerDf)
```

```python
customerDf = df[["Customer ID", "Country"]].drop_duplicates(subset="Customer ID")
customerDf.to_csv("customers.csv", index=False)
print(customerDf)
```

```python
#items
itemDf = pd.DataFrame(
    columns=[
        "StockCode",
        "ItemDescription",
        "Price",        
    ]
)
print(itemDf)
```

```python
itemDf = df[["StockCode", "Description", "Price"]].drop_duplicates(subset = "StockCode")
itemDf.sort_values(by="StockCode", inplace=True)
itemDf.reset_index(drop=True, inplace = True)

# Make the stock_count for each StockCode
Stock = df[["StockCode", "Quantity"]]
Stock.rename(columns={"Quantity": "StockAvailable"}, inplace=True)
stockAvailable = Stock.groupby("StockCode")["StockAvailable"].sum().reset_index()
for i in stockAvailable.index:
    randomNum = random.randint(0, 10000)
    stockAvailable.loc[i, "StockAvailable"] += randomNum
itemDf["StockAvailable"] = stockAvailable["StockAvailable"] # Append the StockAvailable col to items table

itemDf.to_csv("items.csv", index=False)
print(itemDf)
```

```python
#invoice
invoiceDf = pd.DataFrame(
    columns=[
        "InvoiceID",
        "InvoiceDate",
        "customerID"
    ]
)

print(invoiceDf)
```

```python
invoiceDf = df[["Invoice", "InvoiceDate", "Customer ID"]].drop_duplicates(subset="Invoice")
invoiceDf.to_csv("invoice.csv", index=False)
print(invoiceDf)
```
