This file is just a demonstration of how we generated the StockAvailable data. These code sample is inserted to the createTable.ipynb in items code block.

```python
import pandas as pd
import random
```

```python
df = pd.read_csv("./clean2.csv")
print(df)
```

```python
Stock = df[["StockCode", "Quantity"]]
Stock.rename(columns={"Quantity": "StockAvailable"}, inplace=True)
print(Stock)
```

```python
stockAvailable = Stock.groupby("StockCode")["StockAvailable"].sum().reset_index()
print(stockAvailable)
```

```python
for i in stockAvailable.index:
    randomNum = random.randint(0, 10000)
    stockAvailable.loc[i, "StockAvailable"] += randomNum
print(stockAvailable)
```
