```python
import pandas as pd
import random
```

```python
df = pd.read_csv("./clean2.csv")
print(df)
```

```python
Stock = df[["StockCode", "Quantity"]]
print(Stock)
```

```python
stockAvailable = Stock.groupby("StockCode")["Quantity"].sum().reset_index()

print(stockAvailable)
```

```python
for i in stockAvailable.index:
    randomNum = random.randint(0, 10000)
    stockAvailable.loc[i, "Quantity"] += randomNum
print(stockAvailable)
```

```python
stockAvailable.to_csv("StockAvailable.csv", index = False)
```
